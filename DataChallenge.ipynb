# Step 1: Loading and preprocessing 

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()

df = pd.read_csv('Data2026/measures.csv', sep=";", decimal=",")
X = df.drop('activity', axis=1).drop('subject', axis=1)
grouping_feature = df['subject']
y = label_encoder.fit_transform(df['activity'])
```

### Previous analysis has shown that several features are extremely correlated. To correct this, we will classify these features into their own sets of "extremely classified features".

```python
feature_classes = {}
threshold = 0.99

for new_feature in X:
    class_to_add_to = False

    for feature_class in feature_classes:
        for feature in feature_classes[feature_class]:
            if abs(X[feature].corr(X[new_feature])) >= threshold:
                class_to_add_to = feature_class
                break
        if class_to_add_to:
            break
    
    if class_to_add_to:
        feature_classes[class_to_add_to].append(new_feature)
    else:
        feature_classes[new_feature] = [new_feature]

print(f"Sets of features that correlate extremely strongly ({threshold})")
for feature_class in feature_classes:
    if len(feature_classes[feature_class]) > 1:
        print(feature_classes[feature_class])
```
