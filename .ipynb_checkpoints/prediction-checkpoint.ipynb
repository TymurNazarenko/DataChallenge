### Previous analysis has shown that several features are extremely correlated. Therefore, we will remove the highly correlated features to ensure their quality

```python
def remove_correlated_features(X, threshold=0.95):
    corr_matrix = X.corr().abs()
    upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
    to_drop = [col for col in upper.columns if any(upper[col] > threshold)]
    print(f"Removing {len(to_drop)} features due to correlation")
    return X.drop(columns=to_drop), to_drop

X_train, dropped = remove_correlated_features(X_train)
X_to_predict = X_to_predict.drop(columns=dropped)
```
